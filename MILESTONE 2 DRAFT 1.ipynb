```python
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", None)
pd.set_option("display.max_colwidth", None)

print("="*80)
print("FINMARK DATA PIPELINE")
print("Milestone 1 - Data Ingestion & Data Quality Assessment")
print("="*80)
```

```python
trend_path = r"C:\Users\lawre\Downloads\trend_report (1).csv"
marketing_path = r"C:\Users\lawre\Downloads\marketing_summary (1).csv"
event_path = r"C:\Users\lawre\Downloads\event_logs (1).csv"

trend = pd.read_csv(trend_path)
marketing = pd.read_csv(marketing_path)
events = pd.read_csv(event_path)

datasets = {
    "Trend Report": trend,
    "Marketing Summary": marketing,
    "Event Logs": events
}

print("Datasets Successfully Loaded")
```

```python
for name, df in datasets.items():

    print("\n")
    print("="*120)
    print(name.upper())
    print("="*120)

    print(df)
```

```python
print("="*80)
print("INGESTION CHECK")
print("="*80)

for name, df in datasets.items():

    print("\nDataset:", name)

    print("Rows:", df.shape[0])
    print("Columns:", df.shape[1])

    print("Memory Usage:",
          round(df.memory_usage(deep=True).sum()/1024,2),
          "KB")

    print("Successfully Loaded:", not df.empty)
```

```python
print("="*80)
print("SCHEMA VALIDATION")
print("="*80)

for name, df in datasets.items():

    print("\n",name)
    print(df.dtypes)

    schema=pd.DataFrame({
        "Column":df.columns,
        "Datatype":df.dtypes.values
    })

    print(schema)
```

```python
print("="*80)
print("COMPLETENESS CHECK")
print("="*80)

for name, df in datasets.items():

    print("\n",name)

    completeness=pd.DataFrame({
        "Missing Values":df.isnull().sum(),
        "Complete %":((1-df.isnull().mean())*100).round(2)
    })

    print(completeness)
```

```python
print("="*80)
print("NULL VALUE ANALYSIS")
print("="*80)

for name, df in datasets.items():

    print("\n",name)

    nulls=df.isnull().sum()

    null_percent=(nulls/len(df))*100

    report=pd.DataFrame({
        "Null Count":nulls,
        "Null Percentage":null_percent.round(2)
    })

    print(report.sort_values("Null Percentage",ascending=False))
```

```python
print("="*80)
print("VALIDITY CHECK")
print("="*80)

for name, df in datasets.items():

    print("\n",name)

    for col in df.columns:

        duplicate=df[col].duplicated().sum()

        print(f"{col}")
        print("Unique Values:",df[col].nunique())
        print("Duplicate Values:",duplicate)

        if df[col].dtype!="object":

            print("Minimum:",df[col].min())
            print("Maximum:",df[col].max())

        print("-"*50)
```

```python
print("="*80)
print("CONSISTENCY CHECK")
print("="*80)

for name, df in datasets.items():

    print("\n",name)

    print("Duplicate Rows:",df.duplicated().sum())

    print("Duplicated Columns:",
          df.columns.duplicated().sum())

    print("Column Names")

    for col in df.columns:
        print(col)
```

```python
print("="*80)
print("DISTRIBUTION ANALYSIS")
print("="*80)

for name, df in datasets.items():

    print("\n",name)

    numeric=df.select_dtypes(include=np.number)

    if len(numeric.columns)==0:

        print("No Numeric Columns")

    else:

        print(numeric.describe().T)
```

```python
print("="*80)
print("ADVANCED SCHEMA VALIDATION")
print("="*80)

for name, df in datasets.items():

    print("\n",name)

    schema=pd.DataFrame({

        "Column":df.columns,

        "Datatype":df.dtypes,

        "Null Count":df.isnull().sum(),

        "Unique Values":df.nunique()

    })

    print(schema)
```

```python
print("="*80)
print("DATA QUALITY SCORE")
print("="*80)

for name, df in datasets.items():

    total=df.size

    missing=df.isnull().sum().sum()

    duplicate=df.duplicated().sum()

    quality=((total-missing)/total)*100

    print("\n",name)

    print("Total Cells:",total)

    print("Missing Cells:",missing)

    print("Duplicate Rows:",duplicate)

    print("Quality Score:",round(quality,2),"%")
```

```python
print("="*120)
print("CRITICAL FINDINGS FOR CTO REPORT")
print("="*120)

for name, df in datasets.items():

    print("\n")
    print("="*80)
    print(name.upper())
    print("="*80)

    print("Total Records :",len(df))
    print("Total Columns :",len(df.columns))

    duplicate_rows=df.duplicated().sum()
    missing=df.isnull().sum().sum()

    print("Duplicate Rows :",duplicate_rows)
    print("Missing Values :",missing)

    print("\nTop Columns with Missing Data")

    print(
        df.isnull()
        .sum()
        .sort_values(ascending=False)
        .head(10)
    )

    print("\nNumeric Summary")

    print(df.describe(include="all").transpose())

    print("\nCTO Assessment")

    if missing==0 and duplicate_rows==0:

        print("✓ Dataset is production-ready.")
        print("✓ No significant data quality issues detected.")
        print("✓ Suitable for downstream ETL processes.")

    else:

        print("⚠ Data Quality Issues Detected")
        print("• Missing values require preprocessing.")
        print("• Duplicate records should be removed.")
        print("• Data validation rules are recommended before loading into the warehouse.")
        print("• Continuous monitoring is advised to maintain data integrity.")
```
